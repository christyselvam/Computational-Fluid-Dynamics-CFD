```python
import numpy as np
import matplotlib.pyplot as plt
```

# Scalar values

```python
# No of grid poins
N = 51

# Domain size
L = 1

# Corresponding grid spacing
h = np.float64(L/(N-1)) 

#velocity
u = np.float64(0.5)
u_2 = np.float64(50)
u_3 = np.float64(5)

# Density
rho = np.float64(1.0)

# Diffusivity
gamma = np.float64(0.1)

# Peclet number
Pe = rho*u*h/gamma
print("Peclet number is: ", Pe)

Pe_2 = rho*u_2*h/gamma
print("Peclet number is: ", Pe_2)

Pe_3 = rho*u_3*h/gamma
print("Peclet number is: ", Pe_3)

# Iteration number
iterations = 0
```

# Temprature initializing

```python
# initializing temprature field
T = np.zeros(N)
T[0] = 100.
T[N-1] = 500.

# Initializing iterated temprature
T_new = np.zeros(N)
T_new[0] = 100.
T_new[N-1] = 500.

# error related variables
epsilon = 1.E-8
numerical_error = np.float64(1.0)
```

```python
plt.figure(10)
```

# Computation for Pe = 0.1 CDS

```python
while numerical_error > epsilon:
    # Computing for all interior points
    for i in range(1, N-1):
        aw = gamma/h + rho*u
        ae = gamma/h - rho*u
        ap = aw + ae
        T_new[i] = (aw*T[i-1] + ae*T[i+1])/ap
           
    # Resetting the numerical error and recalculate
    numerical_error = 0
    for i in range(1, N-1):
        numerical_error = numerical_error + abs(T[i]-T_new[i])
        
    # Iteration advancement and reassignment
    iterations = iterations + 1
    T = T_new.copy()

    # PLOTTING NUMERICAL ERROR
    if iterations % 500 == 0:
        plt.figure(10)
        plt.semilogy(iterations, numerical_error, 'ko')
        plt.pause(0.01)
            
    
# Defining the position vector from indices
x_dom = np.arange(N)*h
    
# Plotting the variation as a contour plot
plt.figure(11)
plt.plot(x_dom, T, 'gx--', linewidth=2, markersize=10)
    
#Displaying gridlines
plt.grid(True, color='k')
    
# Labelling and providing a title to the plot
plt.xlabel("Position", size=20)
plt.ylabel("Temperature", size=20)
plt.title("T(x)")
    
# Showing the plot on screen
plt.show()
    
```

# Computation for Pe = 10 UPS

```python
while numerical_error > epsilon:
    # Computing for all interior points
    for i in range(1, N-1):
        aw = gamma/h + max(rho*u_2,0)
        ae = gamma/h - max(-rho*u_2,0)
        ap = aw + ae
        T_new[i] = (aw*T[i-1] + ae*T[i+1])/ap
           
    # Resetting the numerical error and recalculate
    numerical_error = 0
    for i in range(1, N-1):
        numerical_error = numerical_error + abs(T[i]-T_new[i])
        
    # Iteration advancement and reassignment
    iterations = iterations + 1
    T = T_new.copy()

    # PLOTTING NUMERICAL ERROR
    if iterations % 500 == 0:
        plt.figure(10)
        plt.semilogy(iterations, numerical_error, 'ko')
        plt.pause(0.01)
            
    
# Defining the position vector from indices
x_dom = np.arange(N)*h
    
# Plotting the variation as a contour plot
plt.figure(11)
plt.plot(x_dom, T, 'gx--', linewidth=2, markersize=10)
    
#Displaying gridlines
plt.grid(True, color='k')
    
# Labelling and providing a title to the plot
plt.xlabel("Position", size=20)
plt.ylabel("Temperature", size=20)
plt.title("T(x)")
    
# Showing the plot on screen
plt.show()
```
