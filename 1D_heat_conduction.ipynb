```python
import numpy as np
import matplotlib.pyplot as plt
```

# No of grid points

```python
N = 11
```

# Domain size

```python
L = 1
```

# Corresponding grid spacing

```python
h = np.float64(L/(N-1)) 
print("Grid spacing is: ", h)
```

# Iteration number

```python
iterations = 0
```

# initializing temprature field

```python
T = np.zeros(N)
print("Initial temperature field is: ", T)
print(len(T))
```

```python
T[N-1] = 1.
print("Temperature field after setting boundary condition is: ", T)
```

# Initializing iterated temprature

```python
T_new = np.zeros(N)
print("Initial iterated temperature field is: ", T_new)
print(len(T_new))
```

```python
T_new[N-1] = 1.
print("Iterated temperature field after setting boundary condition is: ", T_new)
```

# error related variables

```python
epsilon = 1.E-8
numerical_error = 1
```

# Checking the error tolerance

```python
while numerical_error > epsilon:
    # Computing for all interior points
    for i in range(1, N-1):
        T_new[i] = 0.5*(T[i-1]+T[i+1])
        
    # Resetting the numerical error and recalculate
    numerical_error = 0
    for i in range(1, N-1):
        numerical_error = numerical_error + abs(T[i]-T_new[i])
        
    # Iteration advancement and reassignment
    iterations = iterations + 1
    T = T_new.copy()
```

```python
print("Final temperature field is: ", T)
```

# Plotting the results

```python
# Defining the position vector fromindices
x_dom = np.arange(N)*h

# Plotting the variation with customization
plt.plot(x_dom, T, 'gx--', linewidth=2, markersize=10)

#Displaying gridlines
plt.grid(True, color='k')

# Labelling and providing a title to the plot
plt.xlabel("Position", size=20)
plt.ylabel("Temperature", size=20)
plt.title("T(x)")

# Showing the plot on screen
plt.show()
```
